```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\reddy\Downloads\Projects\HeartProblemAnalysis\heart.csv")  # Ensure heart.csv is in the same folder
```

```python
print("📊 Dataset Information:")
print(df.info())
```

```python
print("\n🔹 First 5 rows of data:")
print(df.head())
```

```python
print("\n📈 Summary Statistics:")
print(df.describe())
```

```python
print("\n❓ Missing Values:")
print(df.isnull().sum())
```

```python
print("\n🔹 Total Patients:", len(df))
print("❤️ Heart Disease Cases:", df['target'].sum())
print("💚 No Heart Disease:", len(df) - df['target'].sum())
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['age'], df['chol'], c=df['target'], cmap='coolwarm')
plt.title("Age vs Cholesterol (Red = Heart Disease)")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()
```

```python
plt.figure(figsize=(5,5))
df['sex'].value_counts().plot(kind='bar', color=['skyblue', 'lightcoral'])
plt.title("Gender Distribution")
plt.xticks([0, 1], ['Female', 'Male'], rotation=0)
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(10,8))
corr = df.corr()
plt.imshow(corr, cmap='coolwarm', interpolation='none')
plt.colorbar()
plt.title("Correlation Heatmap")
plt.show()
```
